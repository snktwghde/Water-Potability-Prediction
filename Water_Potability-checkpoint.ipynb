```python
#finding IQR
percentile_25=df['Solids'].quantile(0.25)
percentile_75=df['Solids'].quantile(0.75)
```

```python
iqr=percentile_75-percentile_25
iqr
```

```python
upper_limit=percentile_75+1.5*iqr
lower_limit=percentile_25-1.5*iqr
```

```python
print('UpperLimit:',upper_limit)
print('LowerLimit:',lower_limit)
```

```python
#finding outliers
df[df['Solids']>upper_limit]
```

```python
df[df['Solids']<lower_limit]
```

```python
#capping method
import numpy as np
df1 = df.copy()

df1['Solids'] = np.where(
    df1['Solids'] > upper_limit, upper_limit,
    np.where(
        df1['Solids'] < lower_limit, lower_limit,
        df1['Solids']
    )
)
```

```python
plt.figure(figsize=(16,8))

plt.subplot(2,2,1)
sns.distplot(df['Solids'])


plt.subplot(2,2,2)
sns.boxplot(x='Solids',data=df)


plt.subplot(2,2,3)
sns.distplot(df1['Solids'])


plt.subplot(2,2,4)
sns.boxplot(x='Solids',data=df1)

plt.show()
```
